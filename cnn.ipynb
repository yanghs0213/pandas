```python
from tensorflow.keras import Sequential
from tensorflow.keras import layers

model = Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Dropout(0.1),
    layers.Conv2D(64, 3, activation='relu'),
    layers.MaxPooling2D(),
    layers.Dropout(0.25),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model.summary()
```

```python
# 모델 학습 설정
model.compile(
    # 정답이 0~9의 정수로 주어지는 다중 분류에 사용
    # 예: 정답 레이블이 7이면 7번 클래스의 예측 확률을 기준으로 손실 계산
    loss='sparse_categorical_crossentropy',

    # Adam: 학습 과정에서 가중치를 효율적으로 업데이트하는 최적화 알고리즘
    optimizer='adam',

    # 정확도(accuracy)를 학습 및 평가 과정에서 확인
    metrics=['accuracy']
)
```

```python
from tensorflow.keras.datasets import mnist

# 1. MNIST 데이터셋 불러오기
# 학습 데이터: X_train(이미지), y_train(정답)
# 테스트 데이터: X_test(이미지), y_test(정답)
(X_train, y_train), (X_test, y_test) = mnist.load_data()

display(X_train[0]) # 첫번째 28*28 이미지, 코랩 사용시 출력창에서 이미지 형태로 확인 가능
display('첫번째 y:', y_train[0]) # 첫번째 정답 : 5
display(X_train[1]) # 두번번째 28*28 이미지
display('두번째 y:', y_train[1]) # 두번째 정답 : 0

display(X_train[99]) 
display('100번째 y:', y_train[99]) # 100번째 정답 : 1
display(X_train[100]) 
display('101 번째 y:', y_train[100]) # 101번째 정답 : 5, 첫번째 5와는 다른 모양의 손글씨 5이다.

# 2. 데이터 형태 확인
display(X_train.shape)  # (60000, 28, 28): 학습 이미지 6만 장
display(y_train.shape)  # (60000,): 학습 이미지의 정답 6만 개


display(X_test.shape)   # (10000, 28, 28): 테스트 이미지 1만 장
display(y_test.shape)   # (10000,): 테스트 이미지의 정답 1만 개
```

```python
# 3. 이미지 형태 변경
# 기존: (이미지 개수, 28, 28)
# 변경: (이미지 개수, 28, 28, 1)
# 마지막 1은 흑백 이미지의 채널 수

X_train = X_train.reshape(-1, 28, 28, 1) / 255.0
X_test = X_test.reshape(-1, 28, 28, 1) / 255.0

# 픽셀값을 0~255에서 0~1 범위로 정규화
# 예: 0 → 0.0, 255 → 1.0
# -1은 이미지 개수에 맞게 자동 계산한다.
```

```python
# 4. 모델 학습
model.fit(
    X_train, y_train,
    batch_size=128,  # 한 번에 128개 이미지로 학습
    epochs=10,       # 전체 학습 데이터를 10회 반복
    verbose=1        # 학습 진행 상황 출력
)
```

```python
# 5. 테스트 데이터로 모델 평가
# 학습에 사용하지 않은 이미지로 성능 확인
# loss : 예측 확률과 정답 사이의 손실, 예측의 오차를 평가
# Accuracy : 정답을 맞힌 비율을 평가
loss, accuracy = model.evaluate(
    X_test, y_test,
    verbose=0
)

display(f'Test loss: {loss}')
display(f'Test accuracy: {accuracy}')
```
