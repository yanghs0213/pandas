```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df)
```

```python
df["이름"]
```

```python
# 새로운 열 = 시리즈 타입
# 리스트 -> 시리즈 변경 후 대입
df["성적"] = ['B', 'A', 'C', 'A+', 'B+']
df
```

```python
df["생년"] = 2000 # 값 -> 같은 값의 시리즈로 변경 후 입력
```

```python
df
```

```python
r = df['점수'] + df['출석']
print(r)
print(type(r))
```

```python
df['총점'] = df['점수'] + df['출석']

display(df)
```

```python
# 수정
# [] : 특정 열을 수정, 조건의 행들
# iloc[], loc[] : 특정위치(행,열)의 값을 수정
```

```python
df["생년"] = 1999 # 1999 스칼라값이 Series[1999, 1999, ...] 동일한 사이즈의 시리즈로 변경 후 대입
```

```python
df
```

```python
df.iloc[2, 4] = "C+"
df
```

```python
df.loc[2, "성적"] = "C"
df
```

```python
df.loc[[2, 4], '출석'] = 88
```

```python
df
```

```python
df.iloc[[2, 4], 3] = 85
```

```python
df
```

```python
df["점수"] > 90
```

```python
df.loc[df["점수"] > 90, "점수"] = 99 # 조건식 매개값 자리는 boolean 시리즈 타입 자리
# df.loc[pd.Series([False, True, False, True, False]), "점수"] = 99
# df.loc[[False, True, False, True, False], "점수"] = 99 리스트 -> 시리즈 자동 형변환
df
```

```python
# 삭제 drop() 메서드 사용하는게 가장 편하다
# 특정위치 값들을 삭제 X -> 값을 공백, NaN, ..., 결측치 변경 O
# 행들을 선택해서 삭제 - 세로: axis=0 행
# 열들을 선택해서 삭제 - 가로: axis=1 열
```

```python
df.drop("생년", axis=1)
df
```

```python
# 열만 추출 데이터셋
# df["gender"] 추출된 결과 아래와 같다.
gender = pd.Series([
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "M"
])
print(gender.describe())
print(gender.unique())
print(gender.nunique())
print(gender.value_counts())
```

```python
# 데이터셋 변경 후 차트를 그려야 한다.
gender.value_counts().plot(kind="bar")
```
