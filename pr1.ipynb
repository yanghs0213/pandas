```python
import pandas as pd

score = pd.Series(
    [78, 85, 82, 90, 88, 92],
    index=["1학년중간", "1학년기말",
           "2학년중간", "2학년기말",
           "3학년중간", "3학년기말"]
)

print(score)
```

```python
# 문제 1. Series 기본 정보

# 다음 정보를 출력하라.

print("데이터 개수: ", score.size) # 데이터 개수
print("shape: ", score.shape) # shape
print("데이터 타입: ", score.dtype) # 데이터 타입
print("인덱스: ", score.index) # 인덱스
print("값: ", score.values) # 값
```

```python
# 문제 2. 학년별 성적 조회

# loc를 이용하여 다음을 출력하라.

print(score.loc["1학년중간"]) # 1학년 중간고사 성적
print(score.loc["2학년기말"]) # 2학년 기말고사 성적
print(score.loc["3학년중간"]) # 3학년 중간고사 성적
```

```python
# 문제 3. iloc를 이용한 조회

# iloc를 이용하여 다음을 출력하라.

print(score.iloc[0]) # 첫 번째 시험 성적
print(score.iloc[5]) # 마지막 시험 성적
print(score.iloc[0:2]) # 1학년 두 시험 성적
print(score.iloc[4:6]) # 3학년 두 시험 성적
```

```python
# 문제 4. 성적 슬라이싱

# iloc를 이용하여 다음을 출력하라.

print(score.iloc[0:3]) # 처음 3개의 시험 성적
print(score.iloc[3:6]) # 마지막 3개의 시험 성적
print(score.iloc[1::2]) # 짝수 번째 위치의 성적
print(score.iloc[::-1]) # 역순으로 성적 출력
```

```python
# 문제 5. 기술통계

# 다음 값을 구하라.

print("평균: ", score.mean()) # 평균
print("최댓값: ", score.max()) # 최댓값
print("최솟값: ", score.min()) # 최솟값
print("중앙값: ", score.median()) # 중앙값
print("표준편차: ", score.std()) # 표준편차
print("분산: ", score.var()) # 분산
print("합계: ", score.sum()) # 합계
print(score.describe()) # 그리고 describe()를 이용하여 전체 기술통계량을 출력하라.
```

```python
# 문제 6. 조건 검색

# 다음 조건에 해당하는 시험을 출력하라.

print(score[score >= 90]) # 90점 이상
print(score[(score >= 80) & (score < 90)]) # 80점 이상 90점 미만
avg = score.mean()
print(score[score > avg]) # 평균보다 높은 성적
print(score[score < 70]) # 70점 미만인 성적
```

```python
# 문제 7. 성적 수정·추가

# 다음 작업을 수행하라.

score.iloc[0] = 80 # 1학년중간 성적을 80점으로 수정
score.iloc[5] = 95 # 3학년기말 성적을 95점으로 수정
score.loc["재수강"] = 88 # 재수강이라는 인덱스를 추가하고 성적 88점을 입력
score = score.drop("재수강") # 재수강 데이터를 삭제
print(score)
```

```python
grade = pd.Series(
    ["B", "A", "B", "A", "B", "A"],
    index=["1학년중간", "1학년기말",
           "2학년중간", "2학년기말",
           "3학년중간", "3학년기말"],
    dtype="category"
)

print(grade)
```

```python
# 문제 8. 범주형 데이터 확인

# 다음 정보를 출력하라.

print(grade.dtype) # 자료형 → category
print(grade.cat.categories) # 범주의 종류 → A, B
print(grade.cat.codes) # 범주를 숫자로 표현 → A=0, B=1

# 그리고 일반적인 문자열 Series와 범주형 Series의 차이를 확인하라.
```

```python
# 문제 9. 범주별 개수 확인

# 각 학점이 몇 번 나타나는지 구하라.
print(grade.value_counts())
```

```python
# 문제 10. 범주형 데이터 수정

# 다음 작업을 수행하라.

grade.iloc[0] = "A" # 1학년중간의 등급을 A로 변경
grade = grade.cat.add_categories(["C"]) # C라는 범주가 없을 경우 범주를 먼저 추가
grade.iloc[2] = "C" # 2학년중간의 등급을 C로 변경
print(grade.value_counts()) # 전체 학점 분포를 다시 출력
```
