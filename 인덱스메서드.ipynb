```python
import pandas as pd
```

```python
df = pd.DataFrame({"이름":["김민준", "이서연", "박지훈"], "점수":[85, 90 ,75]}, index=["가", "나", "다"])
# index 매개값 생략하면 Integer 위치값(iloc)과 동일한 index값을 기본값 생성
df # 기본값 display(df)
```

```python
df.columns # 열인덱스
```

```python
df.index # 행인덱스
```

```python
df.index.to_list()
```

```python
df.index.name
```

```python
df.index.name = "index"
df
```

```python
df.index.name = "번호"
df
```

```python
df.index = [101, 200, 333]
df
```

```python
# 인덱스 -> 열로 변경
df2 = df.reset_index(names="번호")
df2
```

```python
# index -> 열
df.reset_index(inplace=True) # 기본값 False 원본 변경 X
df
```

```python
# 열 -> 인덱스 set_index
df.set_index("index", inplace=True)
df
```
