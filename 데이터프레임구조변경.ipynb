```python
import pandas as pd

df = pd.DataFrame({
    "이름": ["철수", "영희"],
    "국어": [80, 90],
    "영어": [70, 95],
    "수학": [90, 85]
})

display(df)
```

```python
df_melt = df.melt(
    id_vars="이름",       # 그대로 유지할 열, 이름은 각 행의 대상을 구분하기 위해 그대로 놔둔다
    var_name="과목",      # 기존 열 이름을 저장할 열
    value_name="점수"     # 기존 열의 값을 저장할 열
)

display(df_melt)
```

```python
df = pd.DataFrame({
    "상품": ["노트북", "모니터", "키보드"],
    "1월": [100, 80, 50],
    "2월": [120, 90, 60],
    "3월": [150, 100, 70]
})

display(df)
```

```python
df_melt = df.melt(
    id_vars="상품",
    var_name="월",
    value_name="매출"
)

display(df_melt)
```

```python
import pandas as pd

df = pd.DataFrame({
    "이름": ["철수", "철수", "철수", "영희", "영희", "영희"],
    "과목": ["국어", "영어", "수학", "국어", "영어", "수학"],
    "점수": [80, 70, 90, 90, 95, 85]
})

display(df)
```

```python
df_pivot = df.pivot(
    index="이름",      # 행으로 사용할 열 -> 
    columns="과목",    # 새로운 열이 될 열 -> 과목열의 값들을 열이름으로
    values="점수"      # 실제 데이터 -> 새로 생성된 열의 값에 사용
)

display(df_pivot)
```

```python
result = df_pivot.reset_index()
display(result)
```

```python
import pandas as pd

df = pd.DataFrame({
    "이름": ["철수", "철수", "철수", "철수", "영희", "영희", "영희"],
    "과목": ["국어", "국어", "영어", "수학", "국어", "영어", "수학"],
    "점수": [80, 50, 70, 90, 90, 95, 85]
})

display(df)
```

```python
# error -> 철수의 국어값이 80, 50 두객의 값이 되어 에러발생 
# ValueError: Index contains duplicate entries, cannot reshape

df_pivot = df.pivot(
    index="이름",      # 행으로 사용할 열
    columns="과목",    # 새로운 열이 될 열
    values="점수"      # 실제 데이터
)

display(df_pivot)
```

```python
# error -> 철수의 국어값이 80, 50 두객의 값이 되어 에러발생 
df_pivot = df.pivot_table(
    index="이름",      # 행으로 사용할 열
    columns="과목",    # 새로운 열이 될 열
    values="점수",      # 실제 데이터
    aggfunc="mean"     # 중복되는 값들에 사용될 집계함수
)

display(df_pivot)
```

```python
df_pivot = df.pivot_table(
    index="이름",       # 행
    columns="과목",     # 열
    values="점수",      # 계산할 값
    aggfunc="sum"       # 중복된 값을 합산
)

display(df_pivot)

# 평균
display(df.pivot_table(
    index="이름",
    columns="과목",
    values="점수",
    aggfunc="mean"
))

# 합계
display(df.pivot_table(
    index="이름",
    columns="과목",
    values="점수",
    aggfunc="sum"
))

# 데이터 개수
display(df.pivot_table(
    index="이름",
    columns="과목",
    values="점수",
    aggfunc="count"
))
```
