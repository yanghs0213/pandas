```python
# 필요한 변수만 선택
df2 = df[["crim", "rm", "dis", "tax", "medv"]]

display(df2.head())
[오전 9:54]# medv 기준으로 등급 생성
# grade 열을 먼저 생성 : 기본값으로 전부 "하"
df2["grade"] = "하"

# medv가 17 이상이면 중
df2.loc[df2["medv"] >= 17, "grade"] = "중"

# medv가 25 이상이면 상
df2.loc[df2["medv"] >= 25, "grade"] = "상"



# category 타입으로 변환
df2["grade"] = df2["grade"].astype("category")

display(df2.head())
display(df2.tail())
display(df2.dtypes)

df2.grade.value_counts()
```
