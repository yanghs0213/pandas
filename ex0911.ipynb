```python
import pandas as pd
import matplotlib.pyplot as plt
# 차트에 한글 깨지지 않도록 차트 폰트변경
plt.rcParams['font.family'] = 'Malgun Gothic'
# 차트에 "-"문자 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 실제 제공되는 데이터 셋
gender = pd.Series([
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "M"
])

s = gender.value_counts() # 전처리
s
```

```python
# 시리즈/DF -> 차트 메서드 실행 -> 축 기본값(변경 가능) -> 레이블 인덱스 -> x축, 값 -> y축
```

```python
s.plot(kind="bar") # s.plot(kind="pie")
```
