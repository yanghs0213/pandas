```python
import pandas as pd
```

```python
# SQL 예
# 집합연산(UNION, ... 행, 열) -> concat 함수
# JOIN(key값(컬럼값) 합친다) -> merge 함수
```

```python
df1 = pd.DataFrame({
    "이름": ["김민수", "이서연"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["박지훈", "최유진"],
    "등급": ["A", "B"]
})

display(df1)
display(df2)
```

```python
df3 = pd.concat([df1, df2], axis=0, ignore_index=True)
df3
```

```python
df4 = pd.concat([df1, df2], axis=1, ignore_index=False)
df4
```

```python
# join 매개변수 ? SQL JOIN과는 다르다
# outer / inner
```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], join="inner")
```

```python
# merge 함수 : SQL JOIN 유사 -> 어려워해야하는데 우리반?
```

```python
import pandas as pd

df_customer = pd.DataFrame({
    "고객ID": [101, 102, 103, 104],
    "이름": ["홍길동", "김철수", "이영희", "박민수"]
})

df_score = pd.DataFrame({
    "고객ID": [101, 102, 103, 105],
    "점수": [90, 80, 95, 70]
})


display(df_customer)
display(df_score)

result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="inner" # inner(기본값), left, right, outer
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="left"
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="right"
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="outer"
)

display(result)
```
