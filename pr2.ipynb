```python
import seaborn as sns
mpg = sns.load_dataset('mpg')
mpg.info()
```

```python
mpg.head()
```

```python
# 문제 1. []로 열 선택

# mpg 데이터프레임에서 mpg 열만 선택하여 출력하시오.
print(mpg["mpg"])
```

```python
# 문제 2. []로 여러 열 선택

# mpg 데이터프레임에서 mpg, weight, origin 세 개의 열을 선택하여 출력하시오.
print(mpg[["mpg", "weight", "origin"]])
```

```python
# 문제 3. iloc로 특정 행 선택

# mpg에서 10번째 행의 데이터를 출력하시오.

# 단, 숫자 위치를 이용하는 iloc을 사용하시오.
print(mpg.iloc[9])
```

```python
# 문제 4. iloc로 행 + 열 선택

# mpg에서 3번째 행의 1번째 열 값을 출력하시오.
print(mpg.iloc[2, 0])
```

```python
# 문제 5. iloc로 행 + 열 슬라이싱

# mpg에서 1~5번째 행과 1~3번째 열을 선택하여 출력하시오.
print(mpg.iloc[0:5, 0:3])
```

```python
# 문제 6. loc로 특정 행 + 열 선택

# mpg에서 인덱스가 10인 행의 mpg 값을 출력하시오.
print(mpg.loc[10, "mpg"])
```

```python
# 문제 7. loc로 여러 행 + 열 선택

# 인덱스가 10~15인 행에서

# - mpg

# - weight

# - origin

# 열만 선택하여 출력하시오.
print(mpg.loc[10:15, ["mpg", "weight", "origin"]])
```

```python
# 문제 8. loc 슬라이싱

# mpg에서 인덱스 10부터 15까지의 데이터를 선택하시오.
print(mpg.loc[10:15])
```

```python
# 문제 9. 조건으로 행 선택

# mpg에서 mpg가 30 이상인 자동차만 선택하시오.
print(mpg[mpg["mpg"] >= 30])
```

```python
# 문제 10. 조건 + 열 선택

# mpg가 30 이상인 자동차의 name과 mpg만 출력하시오.
print(mpg.loc[mpg["mpg"] >= 30, ["name", "mpg"]])
```

```python
# 문제 11. 두 조건으로 행 선택

# mpg가 30 이상이고 weight가 2000 이하인 자동차만 출력하시오.
print(mpg[(mpg["mpg"] >= 30) & (mpg["weight"] <= 2000)])
```
